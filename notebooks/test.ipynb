```python
import numpy as np

def compute_lookat_rot_opengl(pos, target=(0.5, 0.0, 0.0), up=(0.0, 0.0, 1.0)):
    pos    = np.array(pos,    dtype=float)
    target = np.array(target, dtype=float)
    up     = np.array(up,     dtype=float)

    fwd = pos - target
    fwd /= np.linalg.norm(fwd)

    right = np.cross(up, fwd)
    right /= np.linalg.norm(right)

    up_new = np.cross(fwd, right)

    R = np.column_stack([right, up_new, fwd])

    trace = R[0,0] + R[1,1] + R[2,2]

    if trace > 0:                                        # w-dominant
        s = 0.5 / np.sqrt(trace + 1.0)
        w = 0.25 / s
        x = (R[2,1] - R[1,2]) * s
        y = (R[0,2] - R[2,0]) * s
        z = (R[1,0] - R[0,1]) * s
    elif R[0,0] > R[1,1] and R[0,0] > R[2,2]:          # x-dominant
        s = 2.0 * np.sqrt(1.0 + R[0,0] - R[1,1] - R[2,2])
        w = (R[2,1] - R[1,2]) / s
        x = 0.25 * s
        y = (R[0,1] + R[1,0]) / s
        z = (R[0,2] + R[2,0]) / s
    elif R[1,1] > R[2,2]:                               # y-dominant
        s = 2.0 * np.sqrt(1.0 + R[1,1] - R[0,0] - R[2,2])
        w = (R[0,2] - R[2,0]) / s
        x = (R[0,1] + R[1,0]) / s
        y = 0.25 * s
        z = (R[1,2] + R[2,1]) / s
    else:                                               # z-dominant
        s = 2.0 * np.sqrt(1.0 + R[2,2] - R[0,0] - R[1,1])
        w = (R[1,0] - R[0,1]) / s
        x = (R[0,2] + R[2,0]) / s
        y = (R[1,2] + R[2,1]) / s
        z = 0.25 * s

    if w < 0:  # canonical form
        x, y, z, w = -x, -y, -z, -w

    return (round(x,4), round(y,4), round(z,4), round(w,4))

poses = [(0.5,-0.8,1), (0.5,0.8,1), (1,-0.8,1), (1,0,1), (1,0.8,1)]
for pos in poses:
    print(pos, "→", compute_lookat_rot_opengl(pos))
```
